### Parameter vs Argument 

$$
\begin{array}{|l|l|}
\hline
\textbf{Term} & \textbf{What it refers to} \\
\hline
\text{Parameter} & \text{The name in the function's } \texttt{def} \text{ line (the placeholder)} \\
\text{Argument} & \text{The actual value you pass in when calling the function} \\
\hline
\end{array}
$$

There are special type of parameters `*args` and `**kwargs` are parameters too — but instead of catching one value, they catch a variable number of leftover values. The *args has the data type tuple while **kwargs has the data type dictionary.The is a way the parameters are to be ordered as one is defining a function.We have the positional and keyword argument.

#### Positional vs keyword arguments

This is about how you pass the argument when calling the function,by position (order) or by name.

Positional: matched by order.Here Python matches purely because of their position in the call. Get the order wrong, and the values land in the wrong parameters.

Keyword: matched by name Here you explicitly say which parameter each value goes to, using parameter=value. Order no longer matters Python matches by name, not position.

You can mix them but positional must come first.

Where *args and **kwargs fit in
*args collects extra positional arguments (no name given, just extra values) into a tuple.
**kwargs collects extra keyword arguments (name=value pairs with no matching parameter) into a dict.

```python

def sum_num(*nums):
    """ I may want to pass several arguments 
    say I want to get the sum of several numbers 
    that don't follow one another."""

    print("Type:",type(nums))
    total=0
    for i in nums:
        total+=i
    return total
    
print(sum_num(23,34,45,67,80,90,100,900,950))
```

## TYPE HINTING IN PRODUCTION PYTHON


What it is:

Type hints are annotations that tell readers (and tools) what type a variable, parameter, or return value is expected to be. They're optional, and Python does not enforce them at runtime the interpreter still runs the code even if you violate a hint.

```python
def add(a: int, b: int) -> int:
    return a + b
```

$$
\begin{array}{|l|l|}
\hline
\textbf{Part} & \textbf{Meaning} \\
\hline
\texttt{a: int} & \text{Parameter } a \text{ is expected to be an } \texttt{int} \\
\texttt{b: int} & \text{Parameter } b \text{ is expected to be an } \texttt{int} \\
\texttt{-> int} & \text{The function is expected to return an } \texttt{int} \\
\hline
\end{array}
$$

```python
def add(a: int, b: int) -> int:
    return a + b

add("3", "5")   # runs fine! returns "35" (string concatenation)
```

Python doesn't raise an error here it still runs. Type hints are documentation and tooling metadata, not a runtime contract. This surprises a lot of people coming from statically typed languages like Java or C++, where this would be a compile error.

o what are they actually for?
$$
\begin{array}{|l|l|}
\hline
\textbf{Benefit} & \textbf{How} \\
\hline
\text{Catch bugs before running} & \text{Static checkers like } \texttt{mypy} \text{ flag mismatches} \\
\text{Better editor support} & \text{Autocomplete, inline errors in VS Code} \\
\text{Self-documenting code} & \text{No need to read the body to know expected types} \\
\text{Easier to maintain} & \text{Team members see intent immediately} \\
\hline
\end{array}
$$

```python
def circle_area(radius: int=7, pi: float = 3.14) -> float:
    return pi * radius**2
circle_area()
```

```python
def discriminant_hint(a:int,b:int,c:int)-> int:
    """This function calculates the discriminant of a quadratic equation ax^2 + bx + c = 0.
    The discriminant is given by the formula: D = b^2 - 4ac"""
    D=b**2 - 4*a*c
    print(f"The discriminant of the quadratic equation {a}x^2 + {b}x + {c} = 0 is: {D}")
    return D
print(discriminant_hint(1, 2, 3))
```

```python
def discriminant_nature(a:int,b:int,c:int)-> str:
    """ This function determines the nature of the roots of a quadratic equation ax^2 + bx + c = 0 
    based on the types of roots that the quadratics equation has.
    The functions checks for the value of the discriminant D=b^2-4ac and returns a string 
    indicating whether the roots are real and different, real and same, or complex."""
    D=discriminant_hint(a,b,c)
    if D>0:
        return "Roots are real and different"
    elif D==0:
        return "Roots are real and repeated"
    else:
        return "Roots are complex"
print(discriminant_nature(1, -3, 2))
print(discriminant_nature(1, 2, 3))
print(discriminant_nature(1, 2, 1))
```

```python
import os

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from IPython.display import display,Markdown

from sklearn.datasets import load_iris
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
def load_iris_data() -> pd.DataFrame:
    """Load the Iris dataset with 'target' (0, 1, 2) and 'species' (name) columns."""
    iris = load_iris()
    iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
    iris_df["target"] = iris.target
    iris_df["species"] = iris_df["target"].map(dict(enumerate(iris.target_names)))
    return iris_df
iris_df = load_iris_data()
iris_df.drop(columns=["target"], inplace=True)
iris_df
```

```python
def exploratory_data_analysis(df: pd.DataFrame) -> None:
    """Print and display EDA summaries for a DataFrame. Does not modify it."""
    print("Shape:", df.shape)

    print("\nInfo:")
    df.info()

    print("\nDescribe:")
    display(df.describe())

    print("\nMissing values:")
    display(df.isnull().sum())

    print("\nDuplicate rows:", df.duplicated().sum())

    print("\nDtypes:")
    display(df.dtypes)


exploratory_data_analysis(iris_df)
iris_df
```

```python
display(Markdown("## Data Cleaning and Preprocessing"))   # renders a heading --- IGNORE ---
```

```python
def clean_data(df: pd.DataFrame) -> pd.DataFrame:
    """Clean the DataFrame by dropping duplicates and handling missing values."""
    df_cleaned = df.drop_duplicates()
    le = LabelEncoder()
    df_cleaned["species"] = le.fit_transform(df_cleaned["species"])
    return df_cleaned
iris_df_cleaned = clean_data(iris_df)
iris_df_cleaned
```

```python
def visualize_data(df: pd.DataFrame) -> None:
    """Visualize the DataFrame using pairplot and heatmap."""
   

    sns.pairplot(df, hue="species")
    plt.show()

    plt.figure(figsize=(10, 8))
    sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
    plt.title("Correlation Heatmap")
    plt.show()
```
